# Seaborn for Data Analytics

A compact, beginner-friendly Seaborn notebook for Google Colab with short theory and practical examples.

## Introduction to Seaborn

Seaborn is a Python data visualization library built on top of Matplotlib. It provides attractive statistical visualizations and works especially well with Pandas DataFrames.


```python
# Import libraries
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

## 1. Load a Built-in Dataset

Seaborn provides several built-in datasets for learning and practice. The tips dataset contains information about restaurant bills, tips, customers, and meal categories.


```python
# Load the built-in tips dataset
df = sns.load_dataset("tips")

# Display first five rows
df
```

## 2. Count Plot

A count plot shows how many observations belong to each category. It is useful for understanding the frequency of categorical values.


```python
sns.countplot(data=df, x="day")

plt.title("Number of Records by Day")
plt.show()
```

## 3. Bar Plot

A bar plot compares a numerical value across categories. Seaborn can automatically calculate an aggregate such as the mean for each category.


```python
sns.barplot(data=df, x="day", y="total_bill")

plt.title("Average Total Bill by Day")
plt.show()
```

## 4. Histogram

A histogram shows the distribution of numerical data by dividing values into intervals called bins.


```python
sns.histplot(data=df, x="total_bill", bins=15)

plt.title("Distribution of Total Bills")
plt.show()
```

## 5. KDE Plot

A KDE (Kernel Density Estimate) plot provides a smooth representation of the distribution of numerical data. It helps us understand where values are concentrated.


```python
sns.kdeplot(data=df, x="total_bill", fill=True)

plt.title("Density of Total Bills")
plt.show()
```

## 6. Box Plot

A box plot summarizes the distribution of numerical data using the median, quartiles, and possible outliers. It is useful for comparing distributions across categories.


```python
sns.boxplot(data=df, x="day", y="total_bill")

plt.title("Total Bill Distribution by Day")
plt.show()
```

## 7. Scatter Plot

A scatter plot shows the relationship between two numerical variables. Seaborn can use a categorical variable such as gender to create separate groups.


```python
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="sex"
)

plt.title("Total Bill vs Tip")
plt.show()
```

## 8. Line Plot

A line plot is useful for showing changes or trends. It can also compare multiple groups using the `hue` parameter.


```python
sns.lineplot(
    data=df,
    x="size",
    y="tip",
    marker="o"
)

plt.title("Tip by Group Size")
plt.show()
```

## 9. Pair Plot

A pair plot creates multiple scatter plots and distributions for numerical columns. It is useful for quickly exploring relationships among several variables.


```python
sns.pairplot(
    df,
    hue="sex"
)

plt.show()
```

## 10. Heatmap

A heatmap represents numerical values using different color intensities. It is commonly used to visualize correlation matrices.


```python
numeric_df = df.select_dtypes(include="number")
corr = numeric_df.corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

## 11. Styling a Seaborn Plot

Seaborn provides built-in themes that improve the appearance of charts. Styles can be changed globally using `sns.set_theme()`.


```python
sns.set_theme(style="whitegrid")

sns.boxplot(
    data=df,
    x="day",
    y="total_bill"
)

plt.title("Styled Box Plot")
plt.show()
```

## 12. Facet Grid

FacetGrid allows the same visualization to be split into multiple smaller plots based on categories. This makes group-wise comparison easier.


```python
g = sns.FacetGrid(
    df,
    col="time"
)

g.map_dataframe(
    sns.scatterplot,
    x="total_bill",
    y="tip"
)

plt.show()
```

## Quick Summary

Important Seaborn plots covered in this notebook:

- **countplot()** → Counts categorical observations
- **barplot()** → Compares aggregated values
- **histplot()** → Shows numerical distribution
- **kdeplot()** → Shows smooth data density
- **boxplot()** → Shows distribution and outliers
- **scatterplot()** → Shows relationships between variables
- **lineplot()** → Shows trends
- **pairplot()** → Explores relationships among multiple variables
- **heatmap()** → Visualizes matrices such as correlations
- **FacetGrid** → Creates grouped visualizations
