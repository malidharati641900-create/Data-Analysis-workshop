# Matplotlib for Data Analytics

A compact, classroom-friendly guide to Matplotlib with short theory and executable examples.

## Introduction to Matplotlib

Matplotlib is a Python library used to create data visualizations. It helps us represent numerical data using charts and graphs such as line plots, bar charts, histograms, and scatter plots.


```python
# Import libraries
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

## 1. Basic Line Plot

A line plot connects data points using lines. It is commonly used to show trends or changes in values, especially over time.


```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)
plt.show()
```

## 2. Title, Labels and Legend

A title describes the chart, while axis labels explain what the X-axis and Y-axis represent. A legend identifies different lines or data series in a chart.


```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 120, 180, 200]

plt.plot(months, sales, label="Sales")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

## 3. Line Style, Markers and Grid

Matplotlib allows us to customize plots using colors, line styles, markers, and grids. These features make charts easier to read and compare.


```python
plt.plot(
    months,
    sales,
    color="blue",
    linestyle="--",
    marker="o",
    linewidth=2
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

## 4. Multiple Lines

Multiple lines can be plotted on the same chart to compare different datasets or categories. The legend helps distinguish between the lines.


```python
sales_2025 = [100, 150, 120, 180, 200]
sales_2026 = [120, 170, 160, 210, 230]

plt.plot(months, sales_2025, marker="o", label="2025")
plt.plot(months, sales_2026, marker="s", label="2026")

plt.title("Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

## 5. Bar Chart

A bar chart represents values using rectangular bars. It is useful for comparing quantities across different categories.


```python
products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [120, 250, 100, 180]

plt.bar(products, sales)
plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.show()
```

## 6. Horizontal Bar Chart

A horizontal bar chart displays categories on the Y-axis and values on the X-axis. It is useful when category names are long or when horizontal comparison is easier to read.


```python
plt.barh(products, sales)
plt.title("Product Sales")
plt.xlabel("Sales")
plt.ylabel("Product")
plt.show()
```

## 7. Scatter Plot

A scatter plot displays individual data points using dots. It is mainly used to study the relationship or possible correlation between two numerical variables.


```python
hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [40, 45, 50, 55, 60, 68, 75, 85]

plt.scatter(hours, marks)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid()
plt.show()
```

## 8. Histogram

A histogram shows the distribution of numerical data. It divides values into intervals called bins and shows how many observations fall within each interval.


```python
marks = [
    45, 50, 52, 55, 56, 60, 62, 65, 65, 68,
    70, 72, 75, 78, 80, 82, 85, 88, 90, 95
]

plt.hist(marks, bins=5, edgecolor="black")
plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```

## 9. Pie Chart

A pie chart represents parts of a whole. Each slice shows the proportion or percentage of a category relative to the total.


```python
labels = ["Python", "SQL", "Excel", "Power BI"]
values = [40, 25, 20, 15]

plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%"
)

plt.title("Data Analytics Skills")
plt.show()
```

## 10. Subplots

Subplots allow multiple charts to be displayed inside a single figure. They are useful when we want to compare different visualizations of the same or related data.


```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.plot(x, y, marker="o")
plt.title("Line Plot")

plt.subplot(1, 2, 2)
plt.bar(x, y)
plt.title("Bar Chart")

plt.tight_layout()
plt.show()
```

## 11. Figure Size and Axis Limits

The figure size controls the width and height of the complete chart. Axis limits control the range of values displayed on the X-axis and Y-axis.


```python
plt.figure(figsize=(8, 5))

plt.plot(x, y, marker="o")

plt.xlim(1, 5)
plt.ylim(0, 35)

plt.title("Customized Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.grid()
plt.show()
```

## 12. Annotation

Annotations are used to add explanatory text to specific points in a chart. They are helpful for highlighting important values or observations.


```python
plt.plot(x, y, marker="s")

plt.annotate(
    "Highest Value",
    xy=(4, 30),
    xytext=(3, 32),
    arrowprops=dict(arrowstyle="->")
)

plt.title("Annotation Example")
plt.show()
```

## 13. Matplotlib with Pandas

Pandas DataFrames are commonly used in data analytics. Matplotlib can directly use DataFrame columns to create visualizations from real datasets.


```python
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Sales": [100, 150, 120, 180, 200]
}

df = pd.DataFrame(data)

plt.plot(df["Month"], df["Sales"], marker="s")
plt.title("Sales from DataFrame")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

## 14. Real-World Data Analytics Example

In data analytics, visualization is used to communicate insights from datasets. For example, a company can visualize the number of employees across different departments.


```python
data = {
    "Department": ["IT", "HR", "Sales", "Finance", "Marketing"],
    "Employees": [120, 50, 100, 60, 80]
}

df = pd.DataFrame(data)

plt.figure(figsize=(8, 5))
plt.bar(df["Department"], df["Employees"])

plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.show()
```

## 15. Saving a Plot

The `savefig()` function saves a visualization as an image file. This is useful when charts need to be included in reports, presentations, dashboards, or documents.


```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [120, 250, 100, 180, 140]


plt.plot(months, sales, marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.savefig(
    "monthly_sales.png")

plt.show()
```

## Quick Summary

Matplotlib provides several important visualization types:

- **Line Plot** → Shows trends over time
- **Bar Chart** → Compares categories
- **Scatter Plot** → Shows relationships between variables
- **Histogram** → Shows data distribution
- **Pie Chart** → Shows proportions
- **Subplots** → Displays multiple charts together
- **savefig()** → Saves a chart as an image
